```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de rutas seguras para el dataset en /data/
directorio_actual = os.getcwd()
ruta_archivo = os.path.join(directorio_actual, 'data', 'dataset_set_D_telecomunicaciones.xlsx') 
# (Nota: si tu notebook está en la misma carpeta que el excel, usa solo 'dataset_set_D_telecomunicaciones.xlsx')
```

```python
# Cargar el dataset oficial
df = pd.read_excel('dataset_set_D_telecomunicaciones.xlsx')

print(f"Dimensiones del dataset: {df.shape[0]} filas y {df.shape[1]} columnas.\n")
print("Primeros registros:")
display(df.head())

print("\nInformación estructural y tipos de datos:")
df.info()
```

## 1. Resumen Estadístico por Tipo de Servicio
A continuación, agrupamos la información para evaluar el comportamiento del tráfico, la latencia y los tickets de soporte según el tipo de servicio (Móvil, Internet Hogar y Empresas).

```python
resumen_servicio = df.groupby('tipo_servicio').agg({
    'trafico_datos_gb': 'mean',
    'latencia_ms': 'mean',
    'tickets_soporte': 'sum',
    'tiempo_resolucion_hr': 'mean'
}).reset_index()

display(resumen_servicio)
```

```python
# Importar la función desde tu archivo utils_telecom.py creado previamente
from utils_telecom import calcular_estadisticas_latencia

stats_latencia = calcular_estadisticas_latencia(df['latencia_ms'])
print("Estadísticas Avanzadas de Latencia (ms):")
print(stats_latencia)
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='region', y='tiempo_resolucion_hr', hue='tipo_servicio', errorbar=None, palette="Blues")
plt.title('Tiempo Promedio de Resolución por Región y Tipo de Servicio')
plt.xlabel('Región')
plt.ylabel('Tiempo de Resolución (horas)')
plt.legend(title='Tipo de Servicio')
plt.tight_layout()
plt.show()
```
