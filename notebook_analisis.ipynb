```python
# ==========================================
# Importación de Librerías del Proyecto
# ==========================================
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración visual global para gráficos
sns.set_theme(style="whitegrid")
print("Librerías importadas correctamente.")
```

## 1. Carga e Inspección Inicial del Dataset
Verificamos las rutas relativas de manera segura y cargamos el dataset oficial de telecomunicaciones para examinar su estructura, dimensiones y tipos de datos.

```python
# Configuración de ruta y carga de datos
directorio_actual = os.getcwd()
# Ajusta la ruta según la ubicación de tu carpeta 'data' o si está en el mismo directorio
ruta_archivo = os.path.join(directorio_actual, 'data', 'dataset_set_D_telecomunicaciones.xlsx')

if not os.path.exists(ruta_archivo):
    # Si se ejecuta en la misma carpeta raíz o de forma directa
    ruta_archivo = 'dataset_set_D_telecomunicaciones.xlsx'

df = pd.read_excel(ruta_archivo)

print(f"Dimensiones del Dataset: {df.shape[0]} filas y {df.shape[1]} columnas.\n")
print("Primeros 5 registros del dataset:")
display(df.head())

print("\nInformación de tipos de datos y nulos:")
df.info()
```

## 2. Resumen Operacional por Tipo de Servicio
Agrupamos el dataset para evaluar métricas clave como el tráfico promedio de datos, la latencia, la suma de tickets de soporte y el tiempo medio de resolución.

```python
resumen_servicio = df.groupby('tipo_servicio').agg({
    'trafico_datos_gb': 'mean',
    'latencia_ms': 'mean',
    'tickets_soporte': 'sum',
    'tiempo_resolucion_hr': 'mean'
}).reset_index()

# Redondear valores para mejor presentación tabular
resumen_servicio = resumen_servicio.round(2)

print("--- RESUMEN POR TIPO DE SERVICIO ---")
display(resumen_servicio)
```

## 3. Análisis Estadístico Avanzado (Latencia)
Utilizamos nuestro módulo externo o funciones estadísticas con NumPy y SciPy para obtener la media, desviación estándar e intervalo de confianza al 95% ($\text{t-Student}$) de la latencia.

```python
# Importación de la función estadística desde el módulo reutilizable
from utilidad_telecom import calcular_estadisticas_latencia

stats_latencia = calcular_estadisticas_latencia(df['latencia_ms'])

print("Resultados Estadísticos de Latencia (ms):")
for clave, valor in stats_latencia.items():
    print(f"- {clave}: {valor}")
```

## 4. Visualización Gráfica del Desempeño Operacional
Generamos gráficos analíticos para evaluar el tiempo de resolución segmentado por región y tipo de servicio.

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    data=df, 
    x='region', 
    y='tiempo_resolucion_hr', 
    hue='tipo_servicio', 
    errorbar=None, 
    palette="Blues"
)

plt.title('Tiempo Promedio de Resolución por Región y Tipo de Servicio', fontsize=12, fontweight='bold')
plt.xlabel('Región', fontsize=10)
plt.ylabel('Tiempo de Resolución Promedio (horas)', fontsize=10)
plt.legend(title='Tipo de Servicio')
plt.tight_layout()

# Mostrar gráfico en el notebook
plt.show()
```
